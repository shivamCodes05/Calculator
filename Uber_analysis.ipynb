```python
df['pickup_datetime'] = pd.to_datetime(df['pickup_datetime'], errors='coerce')
df = df[df['fare_amount'] > 0]
df = df[(df['passenger_count'] >= 1) & (df['passenger_count'] <= 6)]
df = df.dropna()
print(f"Clean data rows: {len(df)}")
```

```python
df['hour']    = df['pickup_datetime'].dt.hour
df['day']     = df['pickup_datetime'].dt.day_name()
df['month']   = df['pickup_datetime'].dt.month
df['year']    = df['pickup_datetime'].dt.year

def time_of_day(hour):
    if 6 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'

df['time_of_day'] = df['hour'].apply(time_of_day)

print(df[['pickup_datetime', 'hour', 'day', 'time_of_day']].head())
```

```python
sns.set_style("darkgrid")
plt.rcParams['figure.figsize'] = (10, 5)
# hourly rides as per the time of day
plt.figure()
df['hour'].value_counts().sort_index().plot(kind='bar', color='blue')
plt.title('Hourly Ride Distribution')
plt.xlabel('Hour of Day')
plt.ylabel('Number of Rides')
plt.tight_layout()
plt.savefig('graph1_hourly.png')
plt.show()
```

```python
# Fare distribution
plt.figure()
df['fare_amount'].hist(bins=50, color='#FF8559', edgecolor='black')
plt.title('Fare Amount Distribution')
plt.xlabel('Fare ($)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('graph2_fare_dist.png')
plt.show()
```

```python
# Weekday vs Weekend rides
plt.figure()
df['day'].value_counts().plot(kind='bar', color='#3cb371')
plt.title('Rides by Day of Week')
plt.xlabel('Day')
plt.ylabel('Number of Rides')
plt.tight_layout()
plt.savefig('graph3_weekday.png')
plt.show()
```

```python
# Passenger count common
plt.figure()
df['passenger_count'].value_counts().sort_index().plot(
    kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Passenger Count Distribution')
plt.tight_layout()
plt.savefig('graph4_passengers.png')
plt.show()
```

```python
#Time of day average fare
plt.figure()
df.groupby('time_of_day')['fare_amount'].mean().plot(
    kind='bar', color='orchid')
plt.title('Average Fare by Time of Day')
plt.xlabel('Time of Day')
plt.ylabel('Avg Fare ($)')
plt.tight_layout()
plt.savefig('graph5_time_fare.png')
plt.show()
```
